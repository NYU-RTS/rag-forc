```python
import pickle
import xml.etree.ElementTree as ET
from urllib.request import urlopen
```

### Get a list of all documentation pages

```python
with urlopen("https://services.rt.nyu.edu/sitemap.xml") as resp:
    root = ET.fromstring(resp.read())

docs = {el.text for el in root.iter() if el.tag.endswith("loc")}

print("registered", len(docs), "documentation pages")
```

### Confirm the HPC getting started guide is in the set

```python
"https://services.rt.nyu.edu/docs/hpc/getting_started/intro/" in docs
```

### Save to a file

```python
with open("docs.pickle", "wb") as f:
    pickle.dump(docs, f)
```
